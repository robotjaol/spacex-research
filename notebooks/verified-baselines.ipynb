# Verified educational baselines

Install the package first. These examples use synthetic inputs and do not model an operational SpaceX vehicle. Outputs are recomputed when the cells run.

```python
from aerolab.models import landing, navigation, orbit, thermal, fault_injection
import numpy as np
```

```python
descent, contact = landing()
print(contact)
assert contact["contact"]
```

```python
nav = navigation(seed=7)
assert np.linalg.eigvalsh(nav["covariance"]).min() >= -1e-10
print("Covariance checks passed")
```

```python
trajectory = orbit()
print("Analytic circular-position error (m):", trajectory["position_error_m"].max())
```

```python
tank = thermal()
print("Sensible heating only, final T (K):", tank["temperature_k"][-1])
```

```python
faults = fault_injection(seed=7)
assert np.all(np.isnan(faults["estimate"][faults["mode"] == 2]))
print("SAFE estimates unavailable as specified")
```
